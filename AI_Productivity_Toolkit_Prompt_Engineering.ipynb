## 1. Zero-Shot Prompting

Zero-shot prompting asks the AI to perform a task without providing examples.

```python
# Zero-Shot Prompting

zero_shot_prompt = """
Classify the following customer feedback as Positive, Negative, or Neutral.

Feedback:
"The delivery was very fast and the product quality is excellent."
"""

print("ZERO-SHOT PROMPT")
print("=" * 40)
print(zero_shot_prompt)
print("Expected classification: Positive")
```

## 2. Few-Shot Prompting

Few-shot prompting provides examples so that the AI can understand the expected pattern before answering.

```python
# Few-Shot Prompting

few_shot_prompt = """
Classify customer feedback.

Example 1:
Feedback: "The product is amazing and arrived early."
Classification: Positive

Example 2:
Feedback: "The product is damaged and delivery was late."
Classification: Negative

Example 3:
Feedback: "The product is okay, nothing special."
Classification: Neutral

Now classify:

Feedback: "The service was excellent and very helpful."
Classification:
"""

print("FEW-SHOT PROMPT")
print("=" * 40)
print(few_shot_prompt)
print("Expected classification: Positive")
```

## 3. Role-Based Prompting

A role-based prompt tells the AI what role it should act as.

A good prompt can also include:
- Role
- Context
- Instructions
- Constraints
- Expected output format

```python
# Role + Context + Instructions + Constraints

role_based_prompt = """
ROLE:
You are a professional career mentor for undergraduate AIML students.

CONTEXT:
The student is a beginner in Artificial Intelligence and Machine Learning
and wants to improve their technical skills for internships.

INSTRUCTIONS:
Create a simple 4-week learning plan.

CONSTRAINTS:
- Keep the plan beginner-friendly.
- Include Python, SQL, Machine Learning and Git/GitHub.
- Give practical tasks.
- Keep the response concise.

OUTPUT FORMAT:
Week 1:
Week 2:
Week 3:
Week 4:
"""

print("ROLE-BASED PROMPT")
print("=" * 50)
print(role_based_prompt)
```

## 5. Text Summarization

Summarization prompts instruct an AI system to reduce long text into
short, clear, and useful information while preserving the important points.

```python
# Text Summarization Prompt

text_to_summarize = """
Artificial Intelligence is a branch of computer science that focuses on
creating systems capable of performing tasks that normally require human
intelligence. These tasks include learning, reasoning, problem solving,
understanding language, recognizing images, and making decisions.

Machine Learning is a major part of Artificial Intelligence. It allows
computers to learn patterns from data and make predictions without being
explicitly programmed for every individual task. Deep Learning is a
specialized area of Machine Learning that uses neural networks with
multiple layers to process complex data.
"""

summarization_prompt = f"""
TASK:
Summarize the following text.

INSTRUCTIONS:
- Keep the summary concise.
- Include only the important points.
- Use simple language.
- Give the result in 3 bullet points.

TEXT:
{text_to_summarize}
"""

print("SUMMARIZATION PROMPT")
print("=" * 50)
print(summarization_prompt)
```

```python
def create_summary_prompt(text):
    return f"""
Summarize the following text in 3 concise bullet points.

Text:
{text}
"""

sample_text = """
Machine Learning enables computers to learn from data and identify
patterns. It is widely used for prediction, classification,
recommendation systems, fraud detection, and many other applications.
"""

print(create_summary_prompt(sample_text))
```

## 6. Text Classification

Classification prompts ask an AI system to assign text to a predefined category.

```python
# Text Classification

def create_classification_prompt(text):
    return f"""
Classify the following customer feedback into exactly one category:

Categories:
- Positive
- Negative
- Neutral

Customer feedback:
"{text}"

Return only the category name.
"""

feedback = "The app is easy to use and the service was excellent."

print("CLASSIFICATION PROMPT")
print("=" * 50)
print(create_classification_prompt(feedback))
print("\nExpected result: Positive")
```

## 7. Structured JSON Output

Structured outputs make AI responses easier for software to process.
JSON is a common format for structured AI responses.

```python
# Structured JSON Prompt

def create_json_prompt(text):
    return f"""
Analyze the following customer feedback.

Feedback:
"{text}"

Return the result in exactly this JSON structure:

{{
    "sentiment": "Positive/Negative/Neutral",
    "category": "Product/Service/Delivery/Other",
    "summary": "Short summary of the feedback"
}}
"""

feedback = "The product arrived quickly and the quality is excellent."

print("STRUCTURED JSON PROMPT")
print("=" * 50)
print(create_json_prompt(feedback))
```

```python
import json

example_output = {
    "sentiment": "Positive",
    "category": "Product",
    "summary": "The customer is happy with the product quality."
}

print(json.dumps(example_output, indent=4))
```

## 8. Content Generation

Content-generation prompts provide clear instructions about the topic,
audience, tone, length, and output format.

```python
# Content Generation Prompt

def create_content_prompt(topic):
    return f"""
ROLE:
You are a professional technical content writer.

TASK:
Write a short LinkedIn post about: {topic}

AUDIENCE:
Students and beginner developers.

CONSTRAINTS:
- Use professional language.
- Keep it under 150 words.
- Include 3 relevant hashtags.
- Do not make unsupported claims.

OUTPUT:
A short LinkedIn post followed by hashtags.
"""

topic = "Learning Prompt Engineering and AI Productivity"

print("CONTENT GENERATION PROMPT")
print("=" * 50)
print(create_content_prompt(topic))
```

## 9. AI Coding & Debugging Assistant

Prompt engineering can also be used to ask an AI system to explain,
identify, and fix programming errors.

```python
# AI Coding and Debugging Prompt

buggy_code = """
numbers = [10, 20, 30, 40]
total = sum(numbers
print("Total:", total)
"""

debug_prompt = f"""
ROLE:
You are a Python debugging assistant.

TASK:
Analyze the following Python code.

CODE:
{buggy_code}

INSTRUCTIONS:
1. Identify the error.
2. Explain why the error occurs.
3. Provide the corrected code.
4. Keep the explanation beginner-friendly.

OUTPUT FORMAT:
Error:
Explanation:
Corrected Code:
"""

print("CODING & DEBUGGING PROMPT")
print("=" * 50)
print(debug_prompt)
```

```python
# Corrected version of the example code

numbers = [10, 20, 30, 40]
total = sum(numbers)

print("Total:", total)
```

## 10. Reusable Prompt Templates

Reusable templates allow the same prompt structure to be used for
different inputs and tasks.

```python
# Reusable Prompt Templates

def summarization_template(text):
    return f"""
TASK: Summarize the text below.

RULES:
- Use simple language.
- Keep it concise.
- Give 3 bullet points.

TEXT:
{text}
"""


def classification_template(text):
    return f"""
TASK: Classify the following text.

CATEGORIES:
Positive, Negative, Neutral

TEXT:
{text}

OUTPUT:
Return only the category.
"""


def content_template(topic, audience):
    return f"""
ROLE: Professional content writer

TASK:
Create content about: {topic}

AUDIENCE:
{audience}

RULES:
- Use clear language.
- Keep it concise.
- Use a professional tone.

OUTPUT:
Provide the final content only.
"""


# Example usage

sample_text = "The new application is simple, fast and easy to use."

print("REUSABLE SUMMARIZATION TEMPLATE")
print("=" * 50)
print(summarization_template(sample_text))

print("\nREUSABLE CLASSIFICATION TEMPLATE")
print("=" * 50)
print(classification_template(sample_text))

print("\nREUSABLE CONTENT TEMPLATE")
print("=" * 50)
print(content_template(
    "Learning Artificial Intelligence",
    "College students"
))
```

## 11. Prompt Evaluation

Prompt evaluation compares prompts based on clarity, specificity,
constraints, expected output, and reusability.

```python
# Prompt Evaluation

prompt_evaluation = {
    "Clarity": 5,
    "Specificity": 5,
    "Instructions": 5,
    "Constraints": 4,
    "Output_Format": 5,
    "Reusability": 5
}

print("PROMPT EVALUATION")
print("=" * 50)

for criterion, score in prompt_evaluation.items():
    print(f"{criterion}: {score}/5")

average_score = sum(prompt_evaluation.values()) / len(prompt_evaluation)

print("\nAverage Score:", round(average_score, 2), "/ 5")
```
